### What is NLP?

NLP is a subfield of Computer Science and Artificial Intelligence that uses Machine Learning to enable computers to understand and communicate human language.

NLP enables computers and digital devices to recognize, understand and generate text and speech by combining `computational linguistics`, `rule-based modeling of human language` together with `statistical modeling`, `machine learning` and `deep learning`.

It's used in everyday life for powering search engines, prompting chatbots for customer service, voice-operating GPS systems and question-answering digital assistant on smartphones. It plays a growing role in enterprise solutions that help streamline and automate business operations, increase employment productivity and simplify business process.

The benefits of NLP include:
- Automation of repetitive tasks
- Improved data analytics and insights
- Enhanced search
- Content generation

Approaches to NLP:

NLP combines the power of `computational linguistics` with machine learning algorithms and deep learning. Computational linguistics uses data science to analyze language and speech. It includes two main types of analytics:
- syntactical analysis
    - it determines meaning of word, phrase or sentence by parsing the syntax of words and applying preprogrammed rules of grammar
- semantical analysis
    - uses syntactic output to draw meaning from the words and interpret their meaning within the sentence structure

Parsing forms:
- Dependency parsing: looks at relationships between words, such as identifying nouns and verbs
- constituency parsing: builds a parse tree: a rooted and ordered representation of the syntactic structure of sentence or string of words

`Self-supervised Learning (SSL)` is useful for supporting NLP because NLP requires large amounts of labeled data to train AI models. Self-supervised approaches are more time and cost effective which is why they replace some or all manually labeled training data.

Three different approaches to NLP include:
- Rules-based NLP:
    - It used simple if-then decision trees and required preprogrammed rules. e.g includes Moviefone, which had rudimentary NLG capabilities and since there was no Machine Learning or AI capability in rules-based NLP, this function was highly limited and not scalable.

- Statistical NLP:
    - Statistical NLP automatically extracts, classifies and labels elements of text and voice data and then assigns a statistical likelihood to each possible meaning of those elements. It introduced essential techniques of mapping language elements such as words and grammatical rules to a vector representation so that language can be modeled by using mathematical methods, including regression or Markov models.

- Deep learning NLP:
    - It can be viewed as an evolution of Statistical NLP, with the difference that it uses neural network models. There are several subcategories of models:
        - Sequence-to-sequence (seq2seq) models: based on RNN, they've mostly been used in Machine translation by converting phrase from one domain to another (e.g. German to English).
        - Transformer models: they use tokenization of language(position of each token words or subwords) and self-attention(capturing dependencies and relationships) to calculate relation of different language parts to one another. It's trained using self-supervised learning on massive text databases, and a landmark in transformer models was Google's bidirectional encoder representations from transformers (BERT), which became and remains the basis of how Google's search engine works.
        - Autoregressive models: they're trained specifically to predict the next word in a sequence, which represents a huge leap forward in the ability to generate text. e.g. GPT, Llama, Claude, etc
        - Foundation models: they're prebuilt and curate foundation models that can speed the launching of NLP effort and build trust in its operation. They support NLP tasks including content generation and insight extraction.
        
        Additionally, they facilitate RAG, a framework for improving quality of response by linking model to external source of knowledge.

        The models also perform Named Entity Recognition (NER) which involves identifying and extracting key information in a text.

### NLP tasks

Several NLP tasks help process human text and voice data ina ways that help computer make sense of what it's ingesting like:

- Coreference resolution:
    - it's the task of identifying if and when two words refer to the same entity. (e.g. 'she' = 'Mary', for idiom in a text 'bear' != animal but a large & hairy person.)

- Named Entity Recognition(NER):
    - NER identifies words or phrases as useful entities. NER identifies "London" as a location or "Maria" as a person's name.

- Part of Speech tagging:
    - a.k.a grammatical tagging
    - is the process of determining which part of speech a word or piece of text is, based on its use and context

- Word sense disambiguation:
    - It's the selection of a word meaning for a word with multiple possible meanings. (e.g. distinguising meaning of the verb "make" in "make the grade" (to achieve) versus "make a bet" (to place)).

### How NLP works

The overview of typical NLP pipeline and its steps are:

- Text preprocessing:
    - NLP prepares raw text for analysis by transforming it into format that helps machines understand more easily. The steps involved are:
        - tokenization: splitting text into smaller units like words, sentences or phrases.
        - lowercasing letters: converting characters to lowercase
        - stemming or lemmatization
        - text cleaning like removal of unwanted elements, punctuation, special characters and numbers.
- Feature extraction:
    - Converting raw text to numerical representations that machines can analyze and interpret. It involves transforming text to structured data using NLP techniques like Bag of Words and TF-IDF. More advanced techniques include word embeddings like Word2Vec or GloVe.
    - contextual embeddings further enhance this by considering the context in which words appear, allowing for richer, more nuanced representations.
- Text analysis:
    - It involves interpreting and extracting meaningful information from text data through computational techniques like POS (part-of-speech) tagging (identifies grammatical roles of words) amd NER (Named Entity Recognition) which detects specific entities like names, locations and dates.
    - NLU (Natural Language Understanding) is a subset of NLP that focuses on analysing meaning behind sentences. NLU enables software to find similar meanings in different sentecnes to process words that have different meanings.
        - Dependency parsing: analyzes grammatical relationships between words to understand sentence structure
        - Sentiment analysis: determines emotional tone of the text, assessing whether it's positive, negative or neutral
        - Topic modeling: identifies underlying themes or topics within a text or across a corpus of documents.
- Model training:
    - During training, model adjusts its parameters to minimize erros and improve its performance. The effectiveness of NLP modeling is continually refined through evaluation, validation and fine-tuning to enhance accuracy and relevance in real-world applications

### Challenges of NLP

NLP comes with potential pitfalls and some of the associated risks might include:

- Biased training
    - The more diverse the users of NLP, the more significant the risk of baisedness.
- Misinterpretation
    - There's a risk of GIGO, and NLP solutions can be confused if spoken input is in an obscure dialect, mumbled, too full of slang, homonyms, idioms, mispronunciations, etc.
- New vocabulary
    - The conventions of grammar can be broken or intentionally broken, which may create complication in understanding.
- Tone of voice
    - Exaggeration for effect, stressing words for importance or sarcasm can be confused by NLP, making the semantic analysis more difficult and less reliable.
