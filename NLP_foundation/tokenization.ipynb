Tokenization is the process of breaking down text into smaller units, often called tokens. Those tokens can be words, characters, or subwords depending on type of tokenizer used. A tokenizer's job is to break the large text into smaller units and asign them unique identifiers (IDs) that the model can understand.

Tokenization is important because it helps in:
- Structuring raw text:
    - breaking text into tokens make it easier to analyze and process
- Preserve meaning:
    - tokens retain context and meaning essential for tasks like sentiment analysis, translation or summarization
- Standarize input:
    - ensures all inputs follow the same numerical format for the model

Types of tokenization are:
- Word-Based tokenization:
    - splits text into individual words.
    - Advantages:
        - easy to implement
        - retains meaningful units (words)
    - Disadvantage:
        - large vocabularies
        - no semantic similarity
        - unknown tokens

- Character-Based tokenization:
    - breaks text into individual characters rather than words.
    - Advantages:
        - smaller vocabularies
        - minimal unknown tokens
    - Disadvantages:
        - higher computational load
        - less semantic meaning

- Sub-word tokenization:
    - finds a middle ground between word and character based approaches
    - it splits rare words into smaller, meaningful subwords while keeping frequent words intact
    - Advantages:
        - efficient vocabularies
        - semantic preservation
        - low unknown tables
    - Common algorithms:
        - Byte Pair Encoding (BPE)
        - WordPiece
        - SentencePiece

How tokenizers work
* Step 1: Tokenization
    - tokenizer splits text into smaller units based on predefined rules

```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")
sequence = "Learning tokenization is essential in NLP."
tokens = tokenizer.tokenize(sequence)
print(tokens)
```

* Step 2: Conversion to Input IDs
    - Each token is mapped to a unique ID from the tokenizer's vocabulary.

```python
ids = tokenizer.convert_tokens_to_ids(tokens)
print(ids)
```

* Decoding
    - The `decode` method reverses the process, converting IDs back into readable text.

```python
decoded_string = tokenizer.decode(ids)
print(decoded_string)
```
