# Logistic Regression

Logistic Regression is a linear model that predicts the class of a given example. A sigmoid function is applied to the output of the linear function in order to get the result as a class, either 0 or 1. A threshold (e.g. 0.5) or decision boundary is used to mark the function value of sigmoid function to either of the classes.

As for the concepts of overfitting and underfitting, Overfitting occurs when our model is able to learn extremely well from the training dataset but isn't able to perform better on the test dataset. This happens because of less available training dataset, because of which the model memorises rather than learning from examples. Other reasons include:
- Model is too complex
- There are too many features
- No regularization
- High variance (Model learns too much from the data, including random noise)

Underfitting on the other is when the model isn't able to learn at all and fails to learn important features. It performs poorly on both training and testing data. Some of the reasons of underfitting are:
- Model is too simple
- Very high regularization
- Not enough training
- Features are weak or missing

The solutions to both overfitting and underfitting are mentioned as follows:

Underfitting:
- Use more complex models
- Add new features and perform feature engineering
- Reduce regularization
- Scale features properly
- Train for more epochs

Overfitting:
- Collect more training examples
- Reduce model complexity
- Use L1/L2 regularization
- Apply dropout
- Use early stopping
- Clean noisy data

ROC-AUC (Receiver Operating Characteristics - Area Under Curve) helps us understand how well a binary classification model works. It helps understand how well the model separates the positive cases from the negative cases at different threshold level.

It shows how good the model is at telling the difference between the two classes by plotting:

- True Positive Rate (TPR): how often the model correctly predicts positive cases a.k.a Sensitivity or Recall
- False Positive Rate (FPR): how often the model incorrectly predicts a negative case as positive
- Specificity: measures the proportion of actual negatives that the model correctly identifies. It's calculated as 1-FPR.

![ROC-AUC curve](./roc_auc.png)

We use ROC-AUC when:
- dataset is balanced and model needs evaluation across all thresholds
- false positives and false negatives are of similar importance.

The model performance with ROC-AUC:
- High AUC (close to 1): The model effectively distinguishes between positive and negative instances.
- Low AUC (close to 0): The model struggles to differentiate between the two classes.
- AUC around 0.5: The model doesn't learn any meaningful patterns i.e. it's doing random guessing.

### Prediction Bias

If you train a model on the dataset with 5% of emails being spam, then on average, 5% of the emails it classifies should also be spam.

If the mean of the ground truth dataset is 0.05 then the mean of the models' prediction should also be 0.05. We call this zero prediction bias.

### Imbalanced Data

Imbalanced data occurs when one class or category in a dataset is disproportionately represented compared to others. It can exist in both binary classification problem and multi-class classification problems.

For example, in a dataset of 10,000 instances where 95% belong to one class (Class 0) and other 5% belong to other (Class 1), if the model only focuses on Class 0, this can mislead model's performance and skew the evaluation metrics.

This is a problem because of the following reasons:
1. Misleading accuracy
1. Poor minority class performance
1. Skewed evaluation metrics
1. Bias in prediction

Some of the reasons why Machine Learning models perform badly on imbalanced data are mentioned as follows:
1. Skewed loss function
1. Majority class domination
1. Difficulty in separating classes

The techniques for handling imbalanced data are mentioned below:
1. Data based solutions
    1. Under-sampling
    1. Over-sampling
    1. SMOTE
1. Algorithm based solutions
    1. Class weighting
    1. Cost-Sensitive Learning
    1. Ensemble methods
1. Tuning based solutions
    1. Threshold tuning
