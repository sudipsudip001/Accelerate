### 1. KNN (K-Nearest Neighbors algorithm)

- It's also known as "lazy-learner algorithm"
- It is a supervised learning algorithm that can be used for both classification and regression problems.

The KNN algorithm:
- For any data point, calculate its distance (Euclidean, Manhattan, etc) to all the pre-available training data points
- Among them, select the top k data points and find their classes
- The data point belongs to the class that has the highest majority among the given group of classes.


K can be calculated by the formula K = N^(1/2)
It generally performs worse in high-dimension data apparently due to the "curse of dimensionality".

### 2. Cross-Validation

It refers to finding the best algorithm for a certain task by making use of data, by dividing it into multiple parts. The main goal is to estimate how well a model generalizes to independent, unseen data. It's best used for model selection, hyperparameter tuning and performance estimation when the dataset is small and an accurate estimate of generalization error is required.

Some of the most common examples of Cross-Validation include K-fold C.V., Leave-one-out C.V.

Cross-Validation can also be used for "tuning parameters" that helps find the best value for the tuning parameters.

### 3. Support Vector Machines

pre-terms to know about:
- margin (shortest distance between observation and threshold)
- maximal margin classifier
    - (use margins with maximum distance)
    - lame if outliers are present
    - solve by bias/variance tradeoff where choosing a threshold allows misclassifications if permitted
- soft margin (allowed misclassification)
    - found by cross validation
    - classifier a.k.a soft margin classifier
- when data is 1D, support vector classifier is a point,
    - when data is 2D, support vector classifier is a line,
        - when data is 3D, support vector classifier is a plane, and so on.
- Maximal Margin classifiers and Support Vector classifiers can't handle some data type so we use SVM for this case.

The main ideas of SVM are:
- start with data at low dimension
- move the data to a higher dimension
- find a support vector classifier that separates the higher dimension data into two groups

Kernel Functions are used to transform data, where polynomial kernel systematically increases dimensions by settgin the degree of the polynomial. The relationships between each pair of observations are used to find the Support Vector Classifiers. The value of d is also found by C.V.

Kernel trick in SVM is when a kernel function only calculates the relationships between each pair of points as if they are in higher dimension; they don't actually do the transformation. It reduces the amount of computation required for SVM by avoiding math that transforms data from lower to higher dimensions.

### 4. Decision Trees

Decision Trees are non-parametric tests and are used for Classification and Regression Tasks and are a.k.a Classification and Regression Trees(CART).

The components are:
- Root node: topmost node
- Decision node: node that split into branches
- Leaf node: terminal node that predicts the outcome
- Branches: paths connecting nodes representing decision or outcomes

Splitting criteria:
- Gini impurity:
    - measures likelihood or incorrect classification
- Entropy:
    - measures randomness or uncertainty in data
- Information gain
    - reduction in entropy after dataset is split

Stopping criteria:
- maximum depth of a tree
- minimum number of samples required to split a node


Ways to handle overfitting in Decision trees:
- Pruning techniques(pre-pruning and post-pruning)
- feature selection and engineering
- ensemble methods like random forest and gradient boosting
- increasing training data

Some of decision tree algorithms are ID3, C4.5, C5.0, CART, etc.

### 5. Ensemble Learning

It's a ML technique that aggregates two or more learners to produce better predictions. Some of the terminologies associated with it are:

- Base learner: individual model/s
- Weak learner: they perform a little better than random guessing
- Strong learners: Achieve excellent predictive performance
- Parallel model ensembling:
    - parallel methods train each base learners in parallel and independent of one another
    - further divided into homogenous and heterogenous.
        - homogenous: they use the same base learning algorithm to produce all component base learners
        - heterogenous: it ensembles use of different algorithms to produce base learners

Some of the ensemble learning techniques are:
1. Bootstrap aggregation (bagging)
1. Boosting
1. Stacking
1. Voting

1. Bootstrap sampling:
    - generate multiple new datasets by sampling with replacement
    - each new dataset is called a bootstrap sample and has the same size as the original dataset

    Random Forest is an extension of bagging method as it utilizes both bagging and feature randomness to create an uncorrelated forest of decision trees. It operates by constructing multiple decision trees during training and outputting the mode of the classes(classification) or mean prediction(regression) of the individual trees.

2. Boosting:
    - combines predictions of several base estimators to improve the overall perforamnce of a model.
    - boosting methods build models sequentially, each trying to correct the errors made by the previous ones.

    Process description:
    * trains a learner on some initial dataset d, the resultant is typicall weak, misclassifying many samples
    * boosting samples instances from initial dataset to create a new dataset, but prioritizes misclassified data instances from first model or learner.
    * a new learner is trained on this new dataset, then a third dataset is compile from it and so on. The process repeats n times to produce n learners
    * all the weights of learners are combined together to produce final predictions.

    The types of boosting methods are:<br>
        1. Adaptive boosting (Adaboost):
            - It weighs model errors. Adaboost adds weights to previous learner's misclassified samples, causing next learner to prioritize those misclassified samples.<br>

        2. Gradient boosting:
            - It uses residual errors while training new learners. Rather than weight misclassified samples, gradient boosting uses residual errors from a previous model to set target predictions for the next model. It builds model sequentially, and each new model corrects the  residual of the previous one. <br>
        
        3. XGBoost (Extreme Gradient Boosting):
            - It's highly efficient and flexible implementation of gradient boosting designed for speed and performance. XGBoost has key features like Regularization (includes L1 & L2 regularization), Tree pruning, Handling missing values, parallelization, scalability, etc.

3. Stacking:
    - A.k.a stacked generalization, is a heterogenous parallel method that exemplifies what is known as meta-learning.
    - involves training multiple base learners and then using their predictions as inputs to a meta-learner which makes the final prediction.
    - stacking specifically trains several base learners from same dataset using a different training algorithm for each learner.
    - each base learner makes predictions on an unseen dataset. These first models are then compiled and used to train a final model, being the meta-model.
    - The idea is that the meta-learner can learn how to best combine the predictions of the base learners.

4. Voting:
    - involves training multiple models independently & combining their predictions using a voting mechanism.
    - it combines the predictions from multiple models by either averaging their outputs or taking a majority vote.
    - Two types of voting are:
        - hard voting: each model votes for a class, the class with majority vote is the final prediction
        - soft voting: each model provides a probability for each class, and the probabilities are averaged to make the final prediction.

1. K-Means Clustering:
- popular centroid based clustering method in unsupervised learning that divides a dataset into K clusters.
- the goal is to minimize the within-cluter variance, referred to as the Sum of Squared Errors(SSE).

    algorithm
    - choose number of clusters k
    - randomly initialize k cluster centroids
    - assign cluter to each data point on proximity using Euclidean distance formula
    - update centroid of clusters
    - repeat steps 3 and 4 until centroids no longer change or the changes are below a predefined threshold.

- The value of k can be chosen using the Elbow method. It's a heuristic used in determining the number of clusters in a dataset for K-means clustering.


1. DBSCAN:
- This algorithm finds the place that are dense with data points and calls those clusters.
- It can discover clusters of arbitrary shapes
- It uses two parameters to determine how clusters are defined:
    - minPts: the minimum number of data points needed to be clustered together for an area to be considered high-density.
    - eps: the distance used to determine if a data point is in the same area as other data points.

    algorithm:
    - select a point.
    - retrieve all points within eps distance from p.
    - if the number of neighbors is greater than or equal to minPts, p is a core point and a new cluster is created. Otherwise, p is marked as noise.
    - If p is a core point, recursively find all density-reachable points from p and add them to the cluster.
    - Repeat the process until all points are processed.


1. Gaussian Mixture Model (GMM) algorithm:
- It can also be referred as probabilistic KMeans because starting point and training process of KMeans and GMM are same but KMeans uses a distance-based approach, and GMM uses a probabilistic approach.

    algorithm:
    - decide the number of clusters for the given dataset.
    - initiate mean, covariance & weight parameter per cluster.
    - Use expectation maximization algorithm to do the following:
        - expectation step (E step): calculate probability of each data point belonging to each distribution, then evaluate the likelihood function using the current estimate for the parameters.
        - maximization step (M step): update the previous mean, covariance, and weight parametrs to maximize the expected likelihood found in E step.
    - repeat these steps until the model converges (i.e. changes in the parameters are below a threshold.)

1. Agglomerative hierarchy clustering:
- typically used in hierarchical data
- builds a tree of clusters so everything is organized from top-down
- a dendogram, a tree-like figure produced by hierarchical clustering, depicts hierarchical relationships between groups
- individual data points are located at the bottom of the dendogram, while large clusters are located at the top
- to generate different number of clusters, the dendogram can be sliced at various heights

### 7. Singular Value Decomposition

The core formula of SVD is A=UΣVᵀ. If A is a matrix m*n consisting of n number of words (n columns) for m number of documents (m rows), then the decomposed matrices have the following meaning:

- The U Matrix: Documents to Topics
    - Maps documents to topic
    - represents how each document relates to each hidden topic
    - if you have 100 documents, the rows of U tell you if document #1 is more about "sports" or "finance"
    - since it's an orthogonal matrix, the columns are independent, which means SVD tries to find topics that don't overlap. "sports" and "cooking" are kept as distinct categories.

- The Vᵀ Matrix: Words to Topics
    - how much each word relates to each topic
    - tells the computer that the words "touchdown", "goal" and "stadium" all belong to "sports" topic
    - it groups features (words) together so you don't have to look at 10,000 individual words; you can just look at 10 main topics.

- The S (or Σ) Matrix: The volume knob
    - the strength or "importance" of each topic
    - SVD sorts topics by importance. The first singular value might represent a massive, clear topic, while the 50th value might just be random "noise" or very niche sub-topics.


"Truncated SVD"
- S lists topics from most important to least important, we can simply "cut off" the tail end of the list. It's called Truncated SVD.
- This is done to:
    - save memory and make the computer run faster.
    - get rid of the "noise" and keep the "signals". This is the foundation of Dimensionality Reduction - taking a huge, complex dataset and shrinking it down to its most important parts without losing the "story" the data is telling.


GEMINI'S SUMMARY:

Imagine you have a giant spreadsheet where every row is a Book and every column is a Word (counting how often that word appears). SVD is like a filter that distills that massive mess into three parts:

    U (The Map): Tells you which Books belong to which Topics (e.g., this book is 80% "Sci-Fi" and 10% "Romance").

    VT (The Dictionary): Tells you which Words define those Topics (e.g., the "Sci-Fi" topic is made of words like robot, space, and laser).

    S (The Ranking): A list that ranks the Topics by how important they are.

Truncated SVD is simply the act of keeping the top most important topics (like the top 10) and throwing away the rest of the "noise." This shrinks your data size significantly while keeping the most important information intact, which is why it's the secret sauce behind things like search engines and recommendation systems.

### 8. Curse of Dimensionality

The curse of dimensionality is one of the most important problems in data science and machine learning. It happens when dealing with high-dimensional spaces. Two significant challenges that arise are data sparsity and issues with distance metrics. These challenges can cause overfitting in machine learning models and make computations more complex. To tackle these challenges, strategies like dimensionality reduction, feature selection, and regularization techniques can be used.

As Dimensions increases:
- distance grows
- but difference shrinks

Imagine:

🎲 Rolling dice

Roll 2 dice:
You can get 2 or 12 → big variation

Roll 100 dice:
You almost always get ~350

Rarely far from it
👉 So:
Total increases (more dice)
But variation becomes relatively tiny

`Law of Large Numbers (LLN)` suggests that as we increase our sample size or the number of dimensions, the average of our observation gets closer to the expected value.

Let’s consider the example of rolling a fair six-sided dice. The expected mean of a roll is 3.5, which is the average of all possible outcomes. Initially, with just a few rolls, like 5 or 10, the average might be significantly different from 3.5 due to randomness. But as we increase the number of rolls to hundreds or thousands, the average roll value gets closer to 3.5. This phenomenon, where the average of many trials aligns with the expected value, shows the essence of the LLN. It demonstrates that while individual outcomes are unpredictable, the average becomes highly predictable over many trials.

The Euclidean distance between two points in an n-dimensional space is calculated by summing the squared differences across each dimension. We can think of each squared difference as a random variable, similar to a roll of a dice. As the number of dimensions (or rolls) increases, the sum of these ‘rolls’ gets closer to an expected value.


* Key Challenges
    * Data Sparsity:
        - When dimensionality increases, the data points that were too close together in lower dimensions become increasingly separated. The growing emptiness is what we refer to as data sparsity.
    * Distance Metric Problems: When Proximity Loses Meaning
        - In classification tasks, where distance measure is crucial for categorizing, these metrics become ineffective due to higher dimensionality, reducing the effectiveness of the algorithm.

Practical Implications:
- Curse of dimensionality problems show up through overfitting. It happens when a complex model learns noise instead of the patterns in data. It's especially true in high-dimensional spaces where there are many features. The model can make false connections and perform poorly when it sees new data. The curse makes it hard to find patterns in big datasets. High-dimensional data is spread out and sparse, so it's challenging for traditional analysis methods to find meaningful insights. Another implication is that processing high-dimensional data takes a lot of computational power and memory because of which algorithms that work well in lower dimensions become much more complex and resource-heavy as the number of dimensions increases. It requires having more powerful hardware or optimizing algorithms to handle the increased computational load efficiently.

Addressing curse of dimensionality:
- Reduce dimenstionality keeping important information (e.g. PCA algorithm)
- Manifold learning (type of dimensionality reduction)
- Feature selection, where relevant features are chosen to improve model performance
- Regularization techniques prevent overfitting by shrinking less important features.

### 9. Principal Component Analysis

It's the process of reducing higher dimensions data into lower dimension using something called Principal Components. The goal is dimensionality reduction by finding the principal components of the data, which are orthogonal vectors that capture the most significant variateions in the data.

PCA is an unsupervised algorithm. By transforming the original variables into new set of uncorrelated variables (principal components), PCA helps simplify the complexity of high-dimensional datasets, making them easier to visualize and analyze. Principal components are linear combinations of the original variables that have the maximum variance compared to other linear combinations.

The mathematical steps involved in PCA are:
- Create dummy dataset for illustration purpose
- Standarize the data
- Compute the covariance matrix
- Compute eigenvalues and eigenvectors
- Sort eigenvalues and eigenvectors
- Form the principal components
- Resulting Transformed Data
