# Intro to Machine Learning

### Machine Learning is a field of study that gives computers ability to learn without being explicitly programmed to.

In classical programming, we were given `Data` and `Rules` and the objective was to produce     `Answers`.

In the case of Machine Learning, we are supposed to make use of `Data` and `Answers` to create `Rules`.

Machine Learning algorithms are divided into three major types:

1. Supervised Learning
1. Unsupervised Learning
1. Others:
    - Semi-Supervised Learning
    - Reinforcement Learning
    - Recommender Systems

## 1. Supervised Learning

It's the category of ML that uses labeled datasets to train algorithms to predict outcomes and recognize patterns. It makes use of Labeled data or ground truth.

It is furthered classified as:
    - Classification algorithms
    - Regression algorithms

a. Classification algorithms:
    The goal in classification is to predict the label (class) of the given input based on training data. The output is discrete value. Some of the examples include Spam Detection, Breast Cancer, Image Classification, etc.

b. Regression algorithms:
    For regression problems the goal is to predict a continuous output variable based on one or more input variables. The output is continuous values. Examples are Housing price prediction, Stock prediction, Sales forecasting, etc.

## 2. Unsupervised Learning

It's the type of ML that learns from data without human supervision. The models are given unlabeled data and allowed to discover patterns and insights from data without any guidance or supervision. Some common algorithms are Clustering and Dimensionality Reduction.

## 3. Others

### i. Semi-supervised Learning
In real-world labeled datasets are expensive and time-consuming to produce. So, semi-supervised learning is used where an algorithm uses a small amount of labeled data to learn the basic rules and a large amount of unlabeled data to understand underlying shape and distribution of information.

Example: Medical imaging where the doctor can only label a few X-rays, but the computer can look at thousands of unlabeled ones to find common patterns.

### ii. Reinforcement Learning (Trial and Error network)
It's about an Agent interacting with the environment to maximize a Reward. It's like training a dog to fetch a stick, using positive and negative reinforcement.

### iii. Recommender Systems
These are the systems designed to predict the "rating" or "preference" a user would give to an item, based on past interactions, or based on users with similar preferences. Two primary ways are Content-based filtering (More of the same, like if you watch scifi movies, the algorithm will recommend you more scifi movies) and another one is Collaborative filtering (Focuses on User behavior, where if user A and user B liked the same five books, algorithm is going to recommend sixth book to user A that user B liked as well).
