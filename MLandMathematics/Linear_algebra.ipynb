# Linear Algebra

### Vectors are simply represented inside square brackets and they represent coordinates in Linear algebra.

### Topics covered:
- Vector addition, Scalar multiplication
- Basis vectors, span
- Linear transformations and matrices
- Matrix multiplication
- Determinant
- Eigen vectors & values

Operations like Vector addition can be performed which adds two vectors to each other. Similarly, if the vector is multiplied by a scalar value, it is known as scalar multiplication.

Scaling increases/decreases the size of vector based on which scalar value is used for it.

### Basis vectors, span

A basis vectors are something that are linearly independent (they shouldn't be a combination of each other) of each others and they combine to span the entire space. Spanning simply means that the combination of vector is able to represent the given space.

### Linear transformation and matrices

A linear transformation T maps a vector x to a new vector T(x) while preserving vector addition and multiplication. 

Each column of a transformation matrix tells where the original basis vectors land after the transformation.

### Matrix multiplication

Matrix multiplication is simply performing one vector operation after another in a given matrix individually.

### Determinant

Determinant represent the scaling factor for area or volume. For example, if the determinant is 2, the transformation doubles the area of any shape.

A negative determinant indicates that the orientation of the space has been flipped.

### Eigen values & Eigen vectors

Eigenvector is a non-zero vector that only scales during a linear transformation; its span remains unchanged. It can, however, flip to the exact opposite direction if eigenvalue is negative.

Eigenvalue is the scalar factor and it represents how much the eigenvector is stretched, compressed, or flipped.

The characterstic equation is:
    Av=λv

One of the most notable use of Eigen values and Eigen vectors is in Google's Page rank algorithm.
