# Distributions

## Probability Distributions

Probability Distributions have list of all outcomes of a random variable along with their corresponding probability values.

Random Variables are set of all possible values from a Random Experiment.

### Why Probability Distributions?

Probability Distribution functions are important when we can't use probability, because in case of random experiment, the number of possible outcomes is large or even infinite, which makes it impractical to list all outcomes & their probability in table.

We go on to use PDF(Probability Density Function) for continuous and PMF(Probability Mass Function) for discrete random variables.

With Probability Mass Functions, Random variables can take only on a finite, countable sets of discrete values. Some of the examples are Binomial Distribution, Poisson Distribution, Geometric Distribution.

With Probability Distribution Functions, Random variables can take on infinite possible outcomes and the probability of any single outcome is zero and probability of an event occuring over a range of outcomes is represented by PDF. The examples include Normal distribution, Exponential distribution, Beta distribution.

### Density Estimation

Density Estimation can be used for purposes like Hypothesis testing, data analysis, visualization. It's useful in areas of ML where it's often used to estimate probability distribution of input data or model likelihood of certain events.

Commonly used techniques include KDE(Kernel Density Estimation) & GMM(Gaussian Mixture Models).

KDE works by creating a kernel, typically a Gaussian distribution around each data point. We take each data point and assume it as the center of a Gaussian kernel. We then create Gaussian curve around the center point that represents density of data around that point. This process is repeated for all the data points & the result is thus combined to create Gaussian kernels to create final density estimate.

### Normal Distribution

The Normal Distribution is a continuous probability distribution that is symmetrical around the mean. It's defined by two parameters:

- Mean: This is the center of the peak.
- Standard Deviation: The spread of the "bells" wings.

The `Central Limit Theorem (CLT)` is the associated mathematical phenomena with Normal Distribution.

The CLT states if you take enough samples from any population, the distribution of sample means will eventually form a Normal distribution. This allows statisticians to make incredibly accurate predictions about a large group even if they only have a small amount of data.

We know exactly where the data is, based on the 68-95-99.7 Rule:

- 68% falls within 1 standard deviation of the mean
- 95% falls within 2 standard deviations
- 99.7% falls within 3 standard deviations

Machine Learning models assume underlying data is normally distributed. Transformind data to a Standard Normal distribution (mean = 0, s.d. = 1) helps gradient descent converge faster.
