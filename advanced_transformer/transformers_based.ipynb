# In-Depth look at Transformer Based Models

BERT, GPT, T5, BART, and XLNet are members of the Transformer. These models leverage either the Transformer's encoder, decoder, or both for language understanding or text generation.

BERT employes the **encoder**, the GPT models on the other hand utilize **decoder** stacks. The variances in architecture and pre-training objective among language models determine whether a model excels in *text generation task* or *language understanding task*.

The **encoder** performs parallel processing on all tokens, allowing for **bidirectional** understanding, while the **autoregressive** output text generation nature of the **decoder** limits it to **left-to-right** processing.

The original Transformer-based models can only learn a fixed-length segment of 512-2048 tokens. When dealing with lengthy texts, learning multiple segments independently can lead to the inability to capture interdependencies between these segments. **Transformer-XL** incorporates a segment-level recurrence mechanism, which caches the **previous segments** in a hidden state to be reused as extended context when processing the **next segment**. The examples of models that use encoder-only, decoder-only or both are:

* Encoder-only Models: BERT
* Decoder-only Models: GPT, XLNet(Transformer-XL, permutation)
* Encoder-Decoder Models: T5, BART.

### Pre-training objective

The language model undergoes two-step process:

1. Pre-training:
    - Establishes decisive foundation for the model by training on huge & diverse text datastes to establish broad and upper-level understanding of natural language patterns in an unsupervised manner.

2. Fine-tuning:
    - To be more specific on downstrem-tasks fine tuning has to be done. However, finetuning on downstream tasks requires creation of carefully prepared datasets with corresponding labels and often involves modifying fine-structure of the model, which demands significant labor force.


The choice of pre-training objectives can significantly impact performance of Transformer-based model and the degree of adaptation required for fine-tuning on specific tasks. Autoregressive (AR) and Autoencoding (AE) are two successful types of objectives.

**Autoregressive** model focus on regenerating text sequences and are skilled in text generation tasks like machine translation, abstractive summarization and question-answering.

**Autoencoding** model aims to reconstruct original text from corrupted text data and excel in language understanding.

### Positional Encoding

Transformer-based models encode positional information by adding position embeddings to token embeddings, ensuring parallel processing while preserving text sequence order.

The original Transformer model used sinusoidal functions of **absolute positions**, with the insight that relation between two tokens in different positions could be expressed as cosine similarity of their positional embeddings.

Transformer-XL introduced **relative positional** embeddings to handle the position encodings in different segments and recognized that distance between each pair of tokens is a more critical factor in self-attention calculations than their absolute positions.

- Absolute Positional Encoding: BERT, GPT, BART
- Relative Positional Encoding: T5, XLNet

### Tokenizer

Due to inflectional forms of english words, **sub-word based tokenization** is a viable option as it breaks down a word into sub-word units to capture it as root, prefix, suffix and other linguistic elements.

For e.g. "tiresome" -> "tire" "some"
         "tired" -> "tire" "d"

LLMs commonly use sub-word tokenization, with two primary methods being **WordPiece** and **Byte-Pair-Encoding(BPE)**. The WordPiece method collects sub-word units by maximizing the likelihood of the vocabulary dataset over all possible n-gram characters. On the other hand, BPE is similar to WordPiece but uses all 256 Unicode characters and thus can include special characters.

- WordPiece: BERT, T5
- Byte-Pair-Encoding(BPE): GPT-2, GPT-3, BART, RoBERTa.

# MODELS

## 1. BERT (Bidirectional Encoder Representations from Transformer)

![BERT model figure](./BERT.png)

BERT was introduced as a **bidirectional** model that exclusively utilizes **encoder** stacks.

- Pre-training Process:
    - objective: denoising [15% of randomly masked tokens (80% of time they're replaced with [MASK] tokens, 10% time they're replaced with random words, 10% of the time they're kept unchanged to maintain overall context understanding)]
    - pretraining includes **Next Sentence Prediction (NSP)** objective to assess model's ability to understand the overall meaning of a sentence
        - involves a binary classification task to predict whether the next sentence is the actual consecutive sentence. 50% of the time the actual consecutive sentence is presented, 50% of time a random sentence from same literature is provided.

- Fine-tuning Process:
    - necessary for individual downstream tasks (4 downstream tasks categorization):
        - Sentence Pair Classification
        - Single Sentence Classification Task
        - SQuAD(Question-Answering)
        - Named Entitiy Tagging

- Variants: RoBERTa, DistilBERT and DistillRoberta

### RoBERTa (Robustly Optimized BERT Pretraining Approach)

Similar to BERT with some modifications:
1. BERT's NSP is removed from RoBERTa's pretraining approach.
2. Randomization masking of 15% of tokens are changing for each pre-training epoch, as opposed too remaining static throughout all training epochs as in BERT.
3. Uses BPE tokenization instead of WordPiece used by BERT.
4. More extensive training data with lengthier sequence segments are trained.

### DistilBERT (Distilled version of BERT)

- Utilizes smaller, lightweight Transformer network as a **student** model to mimic soft labels or probability distributions of **BERT, a larger teacher model**
- Results in 40% reduction in size of BERT model, with half of encoder stacks being reduced in DistilBERT.
- Still retains 97% of language understanding of BERT, and inference time is improved by 60%.

### DistilRoBERTa (Distilled version of RoBERTa)

- similar to DistilBERT, fundamental concept is to train a smaller and more lightweight student model to emulate behavior of a larger teacher model.

## GPT (Generative Pretrained Transformers)

![GPT figure](./GPT.png)

GPT 1-4 models used decoder stacks only, making them left-right autoregressive models. Though they don't have bidirectional understanding as BERT, their generative approach aligns well with downstream tasks, as all natural language tasks can be treated as generative tasks.

### GPT-1 (117 million parameters)
- relied heavily on fine-tuning for individual finetuning tasks
- didn't exhibit embergent capabilities

### GPT-2 (1.5 billion parameters)
- introduced in context learning for few tasks
- improved its tokenizer by using BPE on top of original spacy tokenizer

### GPT-3 (175 billion parameters)
- strong in-context learning, including zero-shot & few-shot learning
- achieves comparable performance with State of the Art (SOTA) models without finetuning using few-shot or zero-shot settings
- To improve performance on multi-tasks, **InstructGPT** method is used, that combines supervised learning of demonstration texts from labelers, along with Reinforcement Learning from Human Feedback (RLHF).


## BART (Bidirectional and Auto-Regressive Transformers)

![BART architecture diagram](./BART.png)

It was proposed by Facebook and greatly resembles T5 in terms of **denoising pre-training objective** and the **encoder-decoder** architecture, with only difference being that 30% of tokens are masked and **sentence permutation** is used in pre-training text.

## T5 (Text-To-Text Transfer Transformer)

![T5 DIAGRAM](./T5.png)

- Proposed by Google, capable of transforming downstream tasks into text generative tasks, even classification problem.
- uses **encoder-decoder** architecture and **denoising objective**
- in pre-training, 15% of tokens are fed into the encoder are randomly masked.
- consecutive masked tokens are replaced by, a sentinel and are treated as a new single token added to the original vocabulary which helps the model learn to predict how many words are missing in the blank
- ***generative capability** is learned by randomly splitting input text in dataset with first part fed into the encoder and second part treated as output to be auto-regressively regenerated.

## XLNet (Transformer-XL generative autoregressive model)

- pre-training objective developed upon classical autoregressive (AR) objective
- new objective maximizes likelihood of expectation of all possible permutations of a sequence
- XLNet overcomes several limitations, including pretrain-finetune discrepancy caused by BERT's denoising objective and left-to-right context learning limitation of the classical AR objective.
- XLNet is based on Transformer-XL, which includes a segment recurrence mechanism to process extra-long context and uses relative positional encoding.
