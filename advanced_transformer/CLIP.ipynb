# CLIP (Contrastive Language-Image Pretraining)

CLIP is a strategy so good that they can be used to make highly specific and performant classifiers without any training data.

A typical image classifier, incrementally adjusts the model based on its errors until it learns to distinguish between the two. This traditional form of supervised learning is perfectly acceptable for many use cases, and is known to perform well in a variety of tasks. However, this strategy is also known to result in highly specialized models which only perform well within the bounds of their initial training.

To resolve this issue of over-specialization, CLIP aproaches classification in a fundamentally different way; by trying to learn association between images and their annotation through contrastive learning.

### CLIP, in a nutshell

Instead of creating a model that predicts if an image belongs to one of some list of classes, we create a model which predicts if an image belongs to some arbitrary caption. The core idea of CLIP is to use captioned image scraped from internet to create a model which can predict if text is compatible with an image or not.

CLIP does this by learning how to encode images and text in such a way that when the text and image encodings are compared to each other, matching images have a high value and non-matching images have a low value. **In essence, the model learns to map images and text into a landscape such that matching pairs are close together, and not matching pairs are far apart.** This strategy of learning to predict if things belong or don't belong together is comonly referred to as contrastive learning.

In CLIP, contrastive learning is done by learning a text encoder and an image encoder, which learns to put an input into some position in vector space. CLIP compares these positions during training and tries to maximize the closeness of positive pairs, and maximize the closeness of negative pairs.

CLIP allows us to do all sorts of things like:

- We can build image classifiers by just asking the model which text, like "a photo of a cat" and "a photo of a dog" are most likely to be associated with an image.
- We can build image search systems which can be used to find image which is most related to input text.
- Image encoder can be used to extract abstract information about an image which is relevant to text.
- Text encoder by itself can be used to extract abstract information about text which is relevant to images.

## The Components of CLIP

### The Text Encoder

The text encoder within CLIP is a standard transformer encoder.

A transformer can be thought of as a system which takes an entire input sequence of words, then re-represents and compares those words to create an abstract, contextualized representation of the entire input.

The self attention mechanism within a transformer is the main mechanism that creates that contextualized representation.

The modification in CLIP is that it results in a vector rather than a matrix, which is meant to represent the entire input sequence. It does this by simply extracting the vector for the last token in the input sequence. Self-attention mechanism is designed to contextualize each input with every other input, so transformer can learn to encode all necessary meaning into a single vector.


### The Image Encoder

Image encoder converts image to a vector that represents the images meaning. From the perspective of CLIP, the end result ends up being a vector which can be thought of as a summary of the input image. This vector, along with the summary vectors for text, are then used in the next section to construct a multi-modal embedding space.

### The Multi-Modal Embedding space, and CLIP training

The idea of summarizing something into an abstract vector is generally referred to as an "embedding". We "embed" things like images and text into a vector as a way to summarise their general meaning.

![general embedding for CLIP](./general_embedding.png)

Embedding Vectors represent the input as some point in high dimensional space. For example, we could imagine creating encoders which embed their input into a vector of length two. These vectors could then be considered as points in a two-dimensional space, and we could draw their positions.

![positional encoding of vectors](./positions.png)

This can be thought of as **Multi-Modal Embedding Space** and we can train CLIP to put these points in spots such that positive pairs are close to each other.

Most common approach to define closeness is cosine similarity. The cosine of the angle between two vectors can be calculated by measuring the angle between them and passing the angle through the cosine function.
