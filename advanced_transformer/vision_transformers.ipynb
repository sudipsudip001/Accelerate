# Vision Transformers (ViT)

The ViT is based on same attention mechanism as transformers. The transformer in NLP uses encoder & decoder attention branch, the ViT only uses an encoder. The output of encoder is then passed to neural network "head" that makes a prediction.

Its drawback is the requirement for large pretraining datasets.
