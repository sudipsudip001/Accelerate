Deep Learning (DL) had its inception in 2010, and has been used steadily since. It's based on artificial neural network which is nothing but a mimic of human brain. The DL model requires large amount of data to learn and make informed decision therefore is a subset of Machine Learning (ML).

Advantages of DL over ML:
- No need for Feature extraction pre-processing step
- Big Data

Examples of ML, DL and AI are given as follows:

- ML
    - Image Recognition
    - Natural Language Processing
    - Speech Recognition
    - Recommendation System
    - Anomaly Detection

- DL
    - Image Classification
    - Natural Language Processing
    - Speech Recognition
    - Object Detection
    - Facial Recognition
    - Chatbots

- AI
    - Virtual Assistants
    - Recommendation Systems
    - Chatbots
    - Autonomous Vehicles
    - Image and Video Analysis
    - Fraud detection
