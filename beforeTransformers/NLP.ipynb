# N-grams

N-grams are contiguous seqences of 'n' items, typically words in the context of NLP. The items can be characters, words, syllables, depends on the granularity desired. The value of 'n' determines the order of the N-gram.

N-grams help capture the contextual information and semantics within a sequence of words, providing a more nuanced understanding of language.

Examples:
- Unigram (1-grams): single words e.g. "cats", "dogs"
- Bigram (2-grams): pair of consecutive words e.g. "natural language", "deep learning"
- Trigram (3-grams): triplets of consecutive words e.g. "machine learning model", "data science approach", etc.
- 4-gram, 5-grams, etc.

### Significances of N-grams in NLP:

1. Capturing Context and Semantics
1. Improving Language Models
1. Enhancing Text Prediction
1. Information Retrieval
1. Feature Extraction

### Applications of N-grams in NLP:

1. Speech Recognition
1. Machine Translation
1. Predictive Text Input
1. Named Entity Recognition
1. Search Engine Algorithms

```python
import nltk
nltk.download('punkt')

from nltk import ngrams
from nltk.tokenize import word_tokenize

# Example sentence
sentence = "N-grams enhance language processing tasks."

# Tokenize the sentence
tokens = word_tokenize(sentence)

# Generate bigrams
bigrams = list(ngrams(tokens, 2))

# Generate trigrams
trigrams = list(ngrams(tokens, 3))

# Print the results
print("Bigrams:", bigrams)
print("Trigrams:", trigrams)
```
