# Final task

```python
the_dict = {
    'name': [],
    'roll': [],
    'address': [],
    'age': []
}
```

```python
for i in range(10):
    print(f"For person {i}")
    name = input(f"Enter name: ")
    the_dict['name'].append(name)
    roll = input(f"Enter roll number: ")
    the_dict['roll'].append(roll)
    address = input(f"Enter address: ")
    the_dict['address'].append(address)
    age = input(f"Enter age: ")
    the_dict['age'].append(age)
```

```python
the_dict
```

```python
df = pd.DataFrame(the_dict)
df.head()
```

```python
df['age'] = df['age'].astype('int')
```

```python
df.loc[df['age'] <= 15, 'category'] = 'Junior'
df.loc[(df['age'] > 15) & (df['age'] <= 25), 'category'] = 'Mid'
df.loc[df['age'] > 25, 'category'] = 'Senior'
df
```
