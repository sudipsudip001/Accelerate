```python
# program to convert roman numeral to an integer

numeral = "XLIX"
symbols = {'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100, 'D': 500, 'M': 1000}

added = symbols[numeral[len(numeral)-1]]
for i in range(len(numeral)-2, -1, -1):
    if symbols[numeral[i+1]] <= symbols[numeral[i]]:
        added += symbols[numeral[i]]
    else:
        added -= symbols[numeral[i]]

print(f"The eqivalence of Roman numeral {numeral} is {added}.")
```

## Assignment 3.2

```python
import time

def timeit(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        squared = func(*args, **kwargs)
        end_time = time.time()
        print(f"The function took: {end_time-start_time} seconds to complete!")
        return squared
    return wrapper

@timeit
def square_the_number(number):
    return number*number
```

```python
square_the_number(4)
```

# Lesson 4: Object Oriented Programming

### Assignment 4.1

```python
# 1. Class rectagne with methods area() and perimeter()

class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth
    
    def area(self):
        return self.length*self.breadth

    def perimeter(self):
        return 2*(self.length+self.breadth)
```

```python
val = Rectangle(2, 4)
print(val.area())
print(val.perimeter())
```

```python
# 2. Maximum amount of water stored

height = [1,8,6,2,5,4,8,3,7]

l, r = 0, len(height)-1
max_water = float('-inf')

while l < r:
    breadth = min(height[l], height[r])
    length = r-l
    area = length*breadth
    if area > max_water:
        max_water = area
    if height[l] < height[r]:
        l += 1
    else:
        r -= 1
print(f"The maximum amount of water that can be stored is: {max_water}.")
```

## Lesson 5: File I/O operations

```python
# 1.a.

import json

with open("students.json", "r") as file:
    students = json.load(file)

for student in students:
    student['percentage'] = sum(student['marks']) / len(student['marks'])

with open("new_students.json", "w") as file:
    json.dump(students, file, indent=4)
```

```python
# 1.b.

with open("new_students.json", "r") as file:
    records = json.load(file)

failed_students, third_division, second_division, first_division, distinction = 0, 0, 0, 0, 0

for i in records:
    if i['percentage'] < 40:
        failed_students += 1
    elif i['percentage'] > 40 and i['percentage'] < 50:
        third_division += 1
    elif i['percentage'] > 50 and i['percentage'] < 65:
        second_division += 1
    elif i['percentage'] > 65 and i['percentage'] < 80:
        first_division += 1
    else:
        distinction += 1

print(f"Failed students: {failed_students}")
print(f"Third division students: {third_division}")
print(f"Second division students: {second_division}")
print(f"First division students: {first_division}")
print(f"Distinction students: {distinction}")
```
