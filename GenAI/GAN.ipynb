# Generative Adversarial Network

### Discriminative vs. Generative models

### Discriminative models:

A discriminative model is an AI model that can determine how to classify objects. Using supervised learning, an AI model learns that items within a labeled data set have similar characteristics and underlying patterns. The model uses those patterns to predict whether new inputs belong to one labeled class or another. Discriminative models are helpful tools for classification and regression tasks. Three applications of discriminative models include **risk assessment**, **fraud detection**, and **customer segmentation**.

Advantages:
- It provides insights helping to make better decisions
- Helps find patterns in data and improve processes like understanding customer sentiment

Disadvantages:
- They can't create new data using underlying patterns and predictions


### Generative models:

A generative model is a type of AI model that can create new items that look similar and follow the same patterns and characteristics of material the model saw during training. These models require a lot of training materials. The uses for generative models are vast, including both personal day-to-day use, business applications, and even medical and biotech research.

Advantages:
- They're versatile and you can use them in multiple ways like generating business documents, creating marketing materials, optimizing code for software development, forming synthetic data, etc.

Disadvantages:
- As the AI model tends to **hallucainate** and give inaccurate information, it creates a problem
- They don't consistently demonstrate the logic to rule out disinformation or nonsense answers.

## Generative vs. discriminative modes: GAN (Generative Adversarial network) models

Generative models have to accomplish more complex task than discriminative models, because they require enough training materials to understand more complex relationships.

For example, a discriminative model may need to understand the difference between a dog and a cat without considering the complexities of what makes a dog. Generative model not only has to understant the difference between a dog and a cat, but also the relationship between where a dog's nose is located compared to its eyes and how dog's tail can vary between breed of dogs.

You can get best of both models with **GAN**.

In GAN, there are two neural networks: one called **generator** and one called **discriminator**. Generator creates realistic-looking "fake" data using training material, Discriminator attempts to detect the fake and categorize it as such.

The two go back and forth to generate an image/material, until the generator creates the material that the discriminator can't detect as fake, thus winning the game.
