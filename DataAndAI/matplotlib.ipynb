## Matplotlib.pyplot

`pyplot` is the submodule for creating interactive plots.

### Two ways of creating plots in matplotlib:

- matplotlib.pyplot.plot() (for simple plots)
- matplotlib.pyplot.XX() (recommended method for more complex plots.)

```python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3, 4])
```

```python
x = [1, 2, 3, 4]
y = [11, 22, 33, 44]
plt.plot(x, y)
```

```python
fig = plt.figure()
ax = fig.add_subplot()
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_axes([1, 1, 1, 1])
ax.plot(x, y)
plt.show()
```

```python
type(fig), type(ax)
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [11, 22, 33, 44]

fig, ax = plt.subplots(figsize=(10, 10))

ax.plot(x, y)
ax.set(title="Sample simple plot", xlabel="x-axis", ylabel="y-axis")

fig.savefig("simple-plot.png")
```

### Most common types of plots using NumPy arrays

```python
import numpy as np

x = np.linspace(0, 10, 100)
x[:10]
```

```python
fig, ax = plt.subplots()
ax.plot(x, x**2)
```

```python
fig, ax = plt.subplots()
ax.scatter(x, np.exp(x))
```

```python
fig, ax = plt.subplots()
ax.scatter(x, np.sin(x))
```

```python
nut_butter_prices = {"Almond butter": 10,
                     "Peanut butter": 8,
                     "Cashew butter": 12}

fig, ax = plt.subplots()
ax.bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax.set(title="Dan's Nut Butter Store", ylabel="Price ($)")
```

```python
fig, ax = plt.subplots()
ax.barh(list(nut_butter_prices.keys()), list(nut_butter_prices.values()))
```

### Figures in multiple axes

```python
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2,
                                             ncols=2,
                                             figsize=(10, 5))

ax1.plot(x, x/2)
ax2.scatter(np.random.random(10), np.random.random(10))
ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax4.hist(np.random.randn(1000))
```

### Working with the data

```python
car_sales = pd.read_csv('car-sales.csv')

car_sales["Price"] = car_sales["Price"].str.replace('[\$\,\.]', '',
                                                    regex=True)
car_sales
```

```python
car_sales["Price"] = car_sales["Price"].str[:-2]
car_sales
```

```python
car_sales["Sales date"] = pd.date_range('1/1/2024', periods=len(car_sales))
car_sales
```

```python
car_sales["Total sales"] = car_sales["Price"].astype(int).cumsum()
car_sales
```

```python
car_sales.plot(x='Sales date', y='Total sales')
```

```python
car_sales["Price"] = car_sales["Price"].astype(str)

car_sales.plot(x="Odometer (KM)", y="Price", kind="scatter")
```

```python
car_sales["Price"] = car_sales["Price"].astype(int)

car_sales.plot(x="Odometer (KM)", y="Price", kind="scatter")
```

## Working with excel sheet

```python
heart_disease = pd.read_csv("heart.xls")
heart_disease.head()
```

```python
heart_disease["age"].plot.hist(bins=50)
```

```python
heart_disease.plot.hist(figsize=(5,20),
                        subplots=True)
```

### More advanced plots

```python
over_50 = heart_disease[heart_disease["age"]>50]
over_50
```

```python
over_50.plot(kind="scatter",
             x="age",
             y="chol",
             c="target",
             figsize=(10, 6))
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

over_50.plot(kind="scatter",
             x="age",
             y="chol",
             c="target",
             ax=ax)

ax.set_xlim([45, 100])
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

scatter = ax.scatter(over_50["age"],
                     over_50["chol"],
                     c=over_50["target"])

ax.set(title="Heart Disease and Cholesterol levels",
       xlabel="Age",
       ylabel="Cholesterol")

ax.legend(*scatter.legend_elements(),
          title="Target")
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

scatter = ax.scatter(over_50["age"],
                     over_50["chol"],
                     c=over_50["target"])

ax.set(title="Heart Disease and Cholesterol Levels",
       xlabel="Age",
       ylabel="Cholesterol")

ax.legend(*scatter.legend_elements(),
          title="Target")

ax.axhline(over_50["chol"].mean(),
           linestyle="--")
```
