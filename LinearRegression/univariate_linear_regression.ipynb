# Linear Regression

Among the two types of supervised learning algorithm works we discussed earlier, Linear Regression deals with predicting estimating the numeric value. One of the most common example of Linear Regression can be seen in predicting the housing prices based on sizes.

We're given a data that maps input variables to the output variables, i.e. it maps the relation between size of house and prices. Since, only one variable is involved, we can also call it as univariate linear regression.

### Cost Function

Then the cost function deals with the difference between the actual value and the predicted value of the model. It's used in finding how accurate our model for the work at hand. Cost functions can also be referred to as errors.

Some of the depictation of Cost function look like a bowl shaped curve in 2D, hammock in 3D.

### Gradient Descent

Gradient Descent is one of the most popular algorithms used in order to minimize the cost functions by finding out the most optimal values of weights and baises.
